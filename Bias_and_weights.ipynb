```python
import numpy as np
import matplotlib.pyplot as plt

def sigmoid(x):
    return 1 / (1 + np.exp(-x))  # Activation function

# Inputs (features)
x = np.array([0.5, 0.8])  # Example input features

# Different weights
weights = [(-1, 1), (-0.5, 0.5), (0.3, -0.6)]

# Bias values
biases = [-1, 0, 1]

# Range of input values for plotting
x_vals = np.linspace(-2, 2, 100)

# Plot sigmoid output with varying weights
plt.figure(figsize=(12, 5))
plt.subplot(1, 2, 1)
for w in weights:
    weighted_sum = np.dot(x, np.array(w)) * x_vals  # Varying input for visualization
    output = sigmoid(weighted_sum)
    plt.plot(x_vals, output, label=f'Weights: {w}')
plt.axvline(0, color='black', linestyle='--', linewidth=0.5)
plt.axhline(0.5, color='blue', linestyle='--', linewidth=0.5)
plt.xlabel('Input Range')
plt.ylabel('Sigmoid Output')
plt.title('Effect of Varying Weights')
plt.legend()
plt.grid()

# Plot sigmoid output with varying biases
plt.subplot(1, 2, 2)
for b in biases:
    weighted_sum = np.dot(x, np.array([0.3, -0.6])) * x_vals + b  # Varying input for visualization
    output = sigmoid(weighted_sum)
    plt.plot(x_vals, output, label=f'Bias: {b}')
plt.axvline(0, color='black', linestyle='--', linewidth=0.5)
plt.axhline(0.5, color='blue', linestyle='--', linewidth=0.5)
plt.xlabel('Input Range')
plt.ylabel('Sigmoid Output')
plt.title('Effect of Varying Bias')
plt.legend()
plt.grid()

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

def sigmoid(x, weight=1, bias=0):
    return 1 / (1 + np.exp(- (weight * x + bias)))

# Generate input values
x = np.linspace(-10, 10, 100)

# Define different weights and biases
weights = [0.5, 1, 2, 3]
biases = [-2, 0, 2, 4]

plt.figure(figsize=(8, 5))
for w in weights:
    for b in biases:
        plt.plot(x, sigmoid(x, weight=w, bias=b), label=f'Weight={w}, Bias={b}')

plt.axvline(0, color='black', linewidth=0.5)
plt.axhline(0.5, color='gray', linestyle='dotted', linewidth=0.8)
plt.xlabel('Input (x)')
plt.ylabel('Sigmoid Output')
plt.title('Sigmoid Function with Various Weights and Biases')
plt.legend()
plt.grid()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

def sigmoid(x, weight=1):
    return 1 / (1 + np.exp(-weight * x))

# Generate input values
x = np.linspace(-10, 10, 100)

# Define different weights
weights = [0.5, 1, 2, 3, 5]

plt.figure(figsize=(8, 5))
for w in weights:
    plt.plot(x, sigmoid(x, weight=w), label=f'Weight={w}')

plt.axvline(0, color='black', linewidth=0.5)
plt.axhline(0.5, color='gray', linestyle='dotted', linewidth=0.8)
plt.xlabel('Input (x)')
plt.ylabel('Sigmoid Output')
plt.title('Sigmoid Function with Different Weights')
plt.legend()
plt.grid()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Define the linear function
def linear_function(x, m):
    return m * x

# Define different values of x
x_values = [np.linspace(-5, 5, 100), np.linspace(-10, 10, 100), np.linspace(-20, 20, 100)]
labels = ['x in [-5, 5]', 'x in [-10, 10]', 'x in [-20, 20]']

# Define a fixed slope value
m = 2

plt.figure(figsize=(8, 5))
for x, label in zip(x_values, labels):
    plt.plot(x, linear_function(x, m), label=label)

plt.axvline(0, color='black', linewidth=0.5)
plt.axhline(0, color='gray', linestyle='dotted', linewidth=0.8)
plt.xlabel('Input (x)')
plt.ylabel('Output (y)')
plt.title('Linear Function y = mx for Different Ranges of x')
plt.legend()
plt.grid()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Define the function and its derivative
def function(x):
    return x**2

def derivative(x):
    return 2*x

# Initialize parameters
x = np.linspace(-5, 5, 100)
x_current = 4  # Starting point
learning_rate = 0.6
iterations = 10

tangent_x = np.linspace(x_current - 1, x_current + 1, 10)
tangent_y = function(x_current) + derivative(x_current) * (tangent_x - x_current)

plt.figure(figsize=(8, 5))
plt.plot(x, function(x), label='Function: y = x^2')
plt.scatter(x_current, function(x_current), color='red', label='Starting Point')
plt.plot(tangent_x, tangent_y, '--', color='orange', label='Tangent Line')

# Perform Gradient Descent
for _ in range(iterations):
    x_next = x_current - learning_rate * derivative(x_current)
    plt.scatter(x_next, function(x_next), color='blue')
    x_current = x_next

plt.xlabel('x')
plt.ylabel('y')
plt.title('Gradient Descent and Tangent Visualization')
plt.legend()
plt.grid()
plt.show()
```
