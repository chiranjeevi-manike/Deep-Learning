```python
import numpy as np
import matplotlib.pyplot as plt

def sigmoid(x):
    return 1 / (1 + np.exp(-x))  # Activation function

# Inputs (features)
x = np.array([0.5, 0.8])  # Example input features

# Different weights
weights = [(-1, 1), (-0.5, 0.5), (0.3, -0.6)]

# Bias values
biases = [-1, 0, 1]

# Range of input values for plotting
x_vals = np.linspace(-2, 2, 100)

# Plot sigmoid output with varying weights
plt.figure(figsize=(12, 5))
plt.subplot(1, 2, 1)
for w in weights:
    weighted_sum = np.dot(x, np.array(w)) * x_vals  # Varying input for visualization
    output = sigmoid(weighted_sum)
    plt.plot(x_vals, output, label=f'Weights: {w}')
plt.axvline(0, color='black', linestyle='--', linewidth=0.5)
plt.axhline(0.5, color='blue', linestyle='--', linewidth=0.5)
plt.xlabel('Input Range')
plt.ylabel('Sigmoid Output')
plt.title('Effect of Varying Weights')
plt.legend()
plt.grid()

# Plot sigmoid output with varying biases
plt.subplot(1, 2, 2)
for b in biases:
    weighted_sum = np.dot(x, np.array([0.3, -0.6])) * x_vals + b  # Varying input for visualization
    output = sigmoid(weighted_sum)
    plt.plot(x_vals, output, label=f'Bias: {b}')
plt.axvline(0, color='black', linestyle='--', linewidth=0.5)
plt.axhline(0.5, color='blue', linestyle='--', linewidth=0.5)
plt.xlabel('Input Range')
plt.ylabel('Sigmoid Output')
plt.title('Effect of Varying Bias')
plt.legend()
plt.grid()

plt.show()
```
