```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

# Define a simple quadratic function and its gradient
def quadratic(x, y):
    return x**2 + y**2

def gradient(x, y):
    return 2*x, 2*y

# Initialize parameters
x, y = 1.8, 1.5  # Starting point
learning_rate = 0.2
num_updates = 5

# Store values for plotting
x_values, y_values, z_values = [x], [y], [quadratic(x, y)]

# Perform gradient descent updates
for _ in range(num_updates):
    grad_x, grad_y = gradient(x, y)
    x -= learning_rate * grad_x
    y -= learning_rate * grad_y
    x_values.append(x)
    y_values.append(y)
    z_values.append(quadratic(x, y))

# Create a meshgrid for the function
x_range = np.linspace(-2, 2, 50)
y_range = np.linspace(-2, 2, 50)
X, Y = np.meshgrid(x_range, y_range)
Z = quadratic(X, Y)

# Plot the function surface
fig = plt.figure(figsize=(12, 6))
ax = fig.add_subplot(121, projection='3d')
ax.plot_surface(X, Y, Z, cmap='viridis', alpha=0.8)
ax.scatter(x_values, y_values, z_values, color='red', marker='o', label='Gradient Updates')
ax.set_title('Gradient Descent on Quadratic Function')
ax.legend()

# Contour plot
ax2 = fig.add_subplot(122)
contour = ax2.contourf(X, Y, Z, levels=50, cmap='viridis')
ax2.plot(x_values, y_values, marker='o', color='red', linestyle='dashed', label='Gradient Updates')
ax2.set_title('Contour Map with Gradient Descent Steps')
ax2.legend()

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

# Define different functions
def quadratic(x, y):
    return x**2 + y**2

def rosenbrock(x, y):
    return (1 - x)**2 + 100 * (y - x**2)**2

def himmelblau(x, y):
    return (x**2 + y - 11)**2 + (x + y**2 - 7)**2

def ackley(x, y):
    return -20 * np.exp(-0.2 * np.sqrt(0.5 * (x**2 + y**2))) - np.exp(0.5 * (np.cos(2 * np.pi * x) + np.cos(2 * np.pi * y))) + np.e + 20

# Create a meshgrid for visualization
x_range = np.linspace(-2, 2, 100)
y_range = np.linspace(-2, 2, 100)
X, Y = np.meshgrid(x_range, y_range)

Z_quadratic = quadratic(X, Y)
Z_rosenbrock = rosenbrock(X, Y)
Z_himmelblau = himmelblau(X, Y)
Z_ackley = ackley(X, Y)

# Plot different functions
fig = plt.figure(figsize=(15, 10))

ax1 = fig.add_subplot(221, projection='3d')
ax1.plot_surface(X, Y, Z_quadratic, cmap='viridis')
ax1.set_title('Quadratic Function: $x^2 + y^2$')

ax2 = fig.add_subplot(222, projection='3d')
ax2.plot_surface(X, Y, Z_rosenbrock, cmap='plasma')
ax2.set_title('Rosenbrock Function')

ax3 = fig.add_subplot(223, projection='3d')
ax3.plot_surface(X, Y, Z_himmelblau, cmap='coolwarm')
ax3.set_title('Himmelblau Function')

ax4 = fig.add_subplot(224, projection='3d')
ax4.plot_surface(X, Y, Z_ackley, cmap='inferno')
ax4.set_title('Ackley Function')

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

# Define real-world functions
def mean_squared_error(x, y):
    return (x - 1)**2 + (y - 2)**2  # Simple MSE example

def logistic_loss(x, y):
    return np.log(1 + np.exp(-(x * 2 + y * 3)))  # Logistic loss approximation

def sharpe_ratio(x, y):
    return (x - 0.05) / (y + 0.1)  # Simplified Sharpe Ratio function

def economic_cost(x, y):
    return x**2 + 2*y**2 - 3*x*y + 4*x - 5*y + 10  # Example cost function in economics

# Create a meshgrid
x_range = np.linspace(-3, 3, 100)
y_range = np.linspace(-3, 3, 100)
X, Y = np.meshgrid(x_range, y_range)

Z_mse = mean_squared_error(X, Y)
Z_logistic = logistic_loss(X, Y)
Z_sharpe = sharpe_ratio(X, Y)
Z_economic = economic_cost(X, Y)

# Plot real-world functions
fig = plt.figure(figsize=(15, 10))

ax1 = fig.add_subplot(221, projection='3d')
ax1.plot_surface(X, Y, Z_mse, cmap='viridis')
ax1.set_title('Mean Squared Error (MSE)')

ax2 = fig.add_subplot(222, projection='3d')
ax2.plot_surface(X, Y, Z_logistic, cmap='plasma')
ax2.set_title('Logistic Loss Function')

ax3 = fig.add_subplot(223, projection='3d')
ax3.plot_surface(X, Y, Z_sharpe, cmap='coolwarm')
ax3.set_title('Sharpe Ratio (Finance)')

ax4 = fig.add_subplot(224, projection='3d')
ax4.plot_surface(X, Y, Z_economic, cmap='inferno')
ax4.set_title('Economic Cost Function')

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

# Define complex real-world functions
def black_scholes(x, y):
    return np.exp(-y) * (x * np.exp(y) - x)

def wave_equation(x, y):
    return np.sin(x) * np.cos(y) * np.exp(-0.1 * (x**2 + y**2))

def neural_network_loss(x, y):
    return np.log(1 + np.exp(-(x**3 - y**2 + 2*x*y)))

def supply_demand_curve(x, y):
    return np.exp(-((x - 2)**2 + (y - 3)**2)) + 1 / (1 + np.exp(-x * y))

# Create a meshgrid
x_range = np.linspace(-3, 3, 100)
y_range = np.linspace(-3, 3, 100)
X, Y = np.meshgrid(x_range, y_range)

Z_black_scholes = black_scholes(X, Y)
Z_wave = wave_equation(X, Y)
Z_nn_loss = neural_network_loss(X, Y)
Z_supply_demand = supply_demand_curve(X, Y)

# Plot complex real-world functions
fig = plt.figure(figsize=(15, 10))

ax1 = fig.add_subplot(221, projection='3d')
ax1.plot_surface(X, Y, Z_black_scholes, cmap='viridis')
ax1.set_title('Black-Scholes Option Pricing')

ax2 = fig.add_subplot(222, projection='3d')
ax2.plot_surface(X, Y, Z_wave, cmap='plasma')
ax2.set_title('Wave Equation in Physics')

ax3 = fig.add_subplot(223, projection='3d')
ax3.plot_surface(X, Y, Z_nn_loss, cmap='coolwarm')
ax3.set_title('Neural Network Loss Landscape')

ax4 = fig.add_subplot(224, projection='3d')
ax4.plot_surface(X, Y, Z_supply_demand, cmap='inferno')
ax4.set_title('Supply-Demand Economic Model')

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

# Define deep learning-related functions
def mse_loss(x, y):
    return (x - 1)**2 + (y - 2)**2  # Mean Squared Error Loss

def cross_entropy_loss(x, y):
    return - (x * np.log(y + 1e-9) + (1 - x) * np.log(1 - y + 1e-9))  # Cross-Entropy Loss

def relu_activation(x, y):
    return np.maximum(0, x**2 - y**2)  # ReLU Activation Surface

def sigmoid_activation(x, y):
    return 1 / (1 + np.exp(-x * y))  # Sigmoid Activation Function

def tanh_activation(x, y):
    return np.tanh(x * y)  # Tanh Activation Function

def softmax_function(x, y):
    return np.exp(x) / (np.exp(x) + np.exp(y))  # Simplified Softmax Output

# Create a meshgrid
x_range = np.linspace(-3, 3, 100)
y_range = np.linspace(-3, 3, 100)
X, Y = np.meshgrid(x_range, y_range)

Z_mse = mse_loss(X, Y)
Z_cross_entropy = cross_entropy_loss(X, Y)
Z_relu = relu_activation(X, Y)
Z_sigmoid = sigmoid_activation(X, Y)
Z_tanh = tanh_activation(X, Y)
Z_softmax = softmax_function(X, Y)

# Plot deep learning functions
fig = plt.figure(figsize=(18, 12))

ax1 = fig.add_subplot(231, projection='3d')
ax1.plot_surface(X, Y, Z_mse, cmap='viridis')
ax1.set_title('MSE Loss')

ax2 = fig.add_subplot(232, projection='3d')
ax2.plot_surface(X, Y, Z_cross_entropy, cmap='plasma')
ax2.set_title('Cross-Entropy Loss')

ax3 = fig.add_subplot(233, projection='3d')
ax3.plot_surface(X, Y, Z_relu, cmap='coolwarm')
ax3.set_title('ReLU Activation')

ax4 = fig.add_subplot(234, projection='3d')
ax4.plot_surface(X, Y, Z_sigmoid, cmap='inferno')
ax4.set_title('Sigmoid Activation')

ax5 = fig.add_subplot(235, projection='3d')
ax5.plot_surface(X, Y, Z_tanh, cmap='cividis')
ax5.set_title('Tanh Activation')

ax6 = fig.add_subplot(236, projection='3d')
ax6.plot_surface(X, Y, Z_softmax, cmap='magma')
ax6.set_title('Softmax Function')

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Define a simple quadratic loss function: f(x) = x^2
# Its gradient: df/dx = 2x
def gradient(x):
    return 2 * x

# Gradient descent function
def gradient_descent(lr, iterations, x_start):
    x = x_start  # Starting point
    history = [x]  # To store values of x during updates
    for _ in range(iterations):
        x = x - lr * gradient(x)
        history.append(x)
    return history

# Parameters
learning_rates = [0.1, 0.5, 1.0]  # Different learning rates
iterations = 10  # Number of iterations
x_start = 5  # Initial value of x

# Run gradient descent for different learning rates
results = {}
for lr in learning_rates:
    results[lr] = gradient_descent(lr, iterations, x_start)

# Plot gradient updates over iterations
plt.figure(figsize=(10, 6))
for lr, values in results.items():
    plt.plot(values, label=f'lr={lr}')
plt.xlabel('Iteration')
plt.ylabel('x Value')
plt.title('Effect of Learning Rate on Gradient Descent')
plt.legend()
plt.grid()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Define a simple quadratic loss function: f(x) = x^2
# Its gradient: df/dx = 2x
def loss_function(x):
    return x ** 2

def gradient(x):
    return 2 * x

# Gradient Descent
def gradient_descent(lr, iterations, x_start):
    x = x_start
    history = [x]
    for _ in range(iterations):
        x -= lr * gradient(x)
        history.append(x)
    return history

# AdaGrad
def adagrad(lr, iterations, x_start, epsilon=1e-8):
    x = x_start
    history = [x]
    grad_squared = 0
    for _ in range(iterations):
        g = gradient(x)
        grad_squared += g ** 2
        x -= (lr / np.sqrt(grad_squared + epsilon)) * g
        history.append(x)
    return history

# RMSProp
def rmsprop(lr, iterations, x_start, beta=0.9, epsilon=1e-8):
    x = x_start
    history = [x]
    grad_squared = 0
    for _ in range(iterations):
        g = gradient(x)
        grad_squared = beta * grad_squared + (1 - beta) * g ** 2
        x -= (lr / np.sqrt(grad_squared + epsilon)) * g
        history.append(x)
    return history

# Adam
def adam(lr, iterations, x_start, beta1=0.9, beta2=0.999, epsilon=1e-8):
    x = x_start
    history = [x]
    m, v = 0, 0
    for t in range(1, iterations + 1):
        g = gradient(x)
        m = beta1 * m + (1 - beta1) * g
        v = beta2 * v + (1 - beta2) * g ** 2
        m_hat = m / (1 - beta1 ** t)
        v_hat = v / (1 - beta2 ** t)
        x -= (lr / (np.sqrt(v_hat) + epsilon)) * m_hat
        history.append(x)
    return history

# Parameters
learning_rate = 0.1
iterations = 20
x_start = 5

# Run optimization algorithms
methods = {
    'Gradient Descent': gradient_descent(learning_rate, iterations, x_start),
    'AdaGrad': adagrad(learning_rate, iterations, x_start),
    'RMSProp': rmsprop(learning_rate, iterations, x_start),
    'Adam': adam(learning_rate, iterations, x_start)
}

# Plot results
plt.figure(figsize=(10, 6))
for method, values in methods.items():
    plt.plot(values, label=method)
plt.xlabel('Iteration')
plt.ylabel('x Value')
plt.title('Effect of Different Optimization Algorithms')
plt.legend()
plt.grid()
plt.show()
```
