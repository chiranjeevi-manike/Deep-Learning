```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

# Define the function f(x, y)
def f(x, y):
    return x**2 + y**2

# Define the gradient of the function
def gradient(x, y):
    df_dx = 2 * x
    df_dy = 2 * y
    return df_dx, df_dy

# Create a mesh grid for plotting
x = np.linspace(-2, 2, 20)
y = np.linspace(-2, 2, 20)
X, Y = np.meshgrid(x, y)
Z = f(X, Y)

# Compute the gradient vectors
U, V = gradient(X, Y)

# Create the 3D figure
fig = plt.figure(figsize=(10, 7))
ax = fig.add_subplot(111, projection='3d')

# Plot the surface
ax.plot_surface(X, Y, Z, cmap='viridis', alpha=0.7)

# Plot the gradient vectors
ax.quiver(X, Y, Z, U, V, np.zeros_like(U), color='r', length=0.2, normalize=True)

# Labels and title
ax.set_xlabel('X')
ax.set_ylabel('Y')
ax.set_zlabel('f(X, Y)')
ax.set_title('Gradient Visualization in 3D Space')

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

# Define a simple function f(x, y)
def f(x, y):
    return x**2 + y**2

# Define the gradient of the function
def gradient(x, y):
    df_dx = 2 * x
    df_dy = 2 * y
    return df_dx, df_dy

# Gradient Descent Algorithm
def gradient_descent(learning_rate, epochs=50):
    x, y = np.random.uniform(-2, 2, 2)  # Initialize randomly
    history = [(x, y, f(x, y))]
    for _ in range(epochs):
        grad_x, grad_y = gradient(x, y)
        x -= learning_rate * grad_x
        y -= learning_rate * grad_y
        history.append((x, y, f(x, y)))
    return np.array(history)

# Create a mesh grid for plotting
x = np.linspace(-2, 2, 100)
y = np.linspace(-2, 2, 100)
X, Y = np.meshgrid(x, y)
Z = f(X, Y)

# Different learning rates
learning_rates = [0.01, 0.1, 0.5]
histories = [gradient_descent(lr) for lr in learning_rates]

# Plot the effect of learning rates
fig, axes = plt.subplots(1, len(learning_rates), figsize=(15, 5))
for ax, history, lr in zip(axes, histories, learning_rates):
    contour = ax.contourf(X, Y, Z, levels=50, cmap='viridis')
    plt.colorbar(contour, ax=ax, label='f(x, y)')
    ax.plot(history[:, 0], history[:, 1], marker='o', color='w', markersize=4, label=f'LR={lr}')
    ax.scatter(0, 0, color='blue', s=100, label='Global Minimum', edgecolors='black')
    ax.set_xlabel('X')
    ax.set_ylabel('Y')
    ax.set_title(f'Gradient Descent (LR={lr})')
    ax.legend()

plt.show()
```
