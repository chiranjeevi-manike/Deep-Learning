```python
import tensorflow as tf
from tensorflow import keras
import numpy as np
import matplotlib.pyplot as plt

# Load MNIST dataset
(x_train, y_train), (x_test, y_test) = keras.datasets.mnist.load_data()

# Normalize data (scale pixel values to [0, 1])
x_train, x_test = x_train / 255.0, x_test / 255.0

# Flatten the images (from 28x28 to 784)
x_train = x_train.reshape(-1, 784)
x_test = x_test.reshape(-1, 784)

# Define activation functions
activations = {
    "ReLU": "relu",
    "Leaky ReLU": keras.layers.LeakyReLU(alpha=0.01),
    "PReLU": keras.layers.PReLU(),
    "ELU": keras.layers.ELU(alpha=1.0),
    "Swish": tf.nn.swish
}

# Function to build and train a model
def train_model(activation_name, activation_fn):
    model = keras.Sequential([
        keras.layers.Dense(256, activation=activation_fn, input_shape=(784,)),
        keras.layers.Dense(128, activation=activation_fn),
        keras.layers.Dense(64, activation=activation_fn),
        keras.layers.Dense(10, activation="softmax")  # Output layer
    ])

    model.compile(optimizer="adam", loss="sparse_categorical_crossentropy", metrics=["accuracy"])

    print(f"\nTraining model with {activation_name} activation...\n")
    history = model.fit(x_train, y_train, epochs=5, batch_size=64, validation_data=(x_test, y_test), verbose=1)

    return history.history["val_accuracy"]

# Train models with different activations
results = {}
for name, act in activations.items():
    results[name] = train_model(name, act)

# Plot the results
plt.figure(figsize=(8, 6))
for name, acc in results.items():
    plt.plot(acc, label=name, marker='o')

plt.xlabel("Epochs")
plt.ylabel("Validation Accuracy")
plt.title("Activation Functions Comparison in DNN")
plt.legend()
plt.grid(True)
plt.show()
```

Input Layer (784)
       ↓
Hidden Layer 1 (256 neurons, Activation: ReLU)
       ↓
Hidden Layer 2 (128 neurons, Activation: ReLU)
       ↓
Hidden Layer 3 (64 neurons, Activation: ReLU)
       ↓
Output Layer (10 neurons, Softmax Activation)

```python
import tensorflow as tf
from tensorflow import keras
import numpy as np
import matplotlib.pyplot as plt
from tensorflow.keras.utils import plot_model

# Load MNIST dataset
(x_train, y_train), (x_test, y_test) = keras.datasets.mnist.load_data()

# Normalize data (scale pixel values to [0, 1])
x_train, x_test = x_train / 255.0, x_test / 255.0

# Flatten images (from 28x28 to 784)
x_train = x_train.reshape(-1, 784)
x_test = x_test.reshape(-1, 784)

# Define activation functions
activations = {
    "ReLU": "relu",
    "Leaky ReLU": keras.layers.LeakyReLU(alpha=0.01),
    "PReLU": keras.layers.PReLU(),
    "ELU": keras.layers.ELU(alpha=1.0),
    "Swish": tf.nn.swish
}

# Function to build and train a model
def train_model(activation_name, activation_fn):
    model = keras.Sequential([
        keras.layers.Dense(256, activation=activation_fn, input_shape=(784,), name="Hidden_Layer_1"),
        keras.layers.Dense(128, activation=activation_fn, name="Hidden_Layer_2"),
        keras.layers.Dense(64, activation=activation_fn, name="Hidden_Layer_3"),
        keras.layers.Dense(10, activation="softmax", name="Output_Layer")
    ])

    model.compile(optimizer="adam", loss="sparse_categorical_crossentropy", metrics=["accuracy"])

    print(f"\n🔹 Training model with {activation_name} activation...\n")
    history = model.fit(x_train, y_train, epochs=5, batch_size=64, validation_data=(x_test, y_test), verbose=1)

    # Save model architecture plot
    plot_model(model, to_file=f"{activation_name}_architecture.png", show_shapes=True, show_layer_names=True)

    return history.history["val_accuracy"]

# Train models with different activations
results = {}
for name, act in activations.items():
    results[name] = train_model(name, act)

# Plot the results
plt.figure(figsize=(8, 6))
for name, acc in results.items():
    plt.plot(acc, label=name, marker='o')

plt.xlabel("Epochs")
plt.ylabel("Validation Accuracy")
plt.title("Activation Functions Comparison in DNN")
plt.legend()
plt.grid(True)
plt.show()
```
