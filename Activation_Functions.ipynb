```python
import numpy as np
import matplotlib.pyplot as plt

def binary_step(x):
    return np.where(x >= 0, 1, 0)

# Generate x values from -10 to 10
x = np.linspace(-10, 10, 100)
y = binary_step(x)

# Plot the Binary Step function
plt.figure(figsize=(8, 5))
plt.plot(x, y, label='Binary Step Function', color='b')
plt.axhline(y=0, color='k', linestyle='--', linewidth=0.5)
plt.axhline(y=1, color='k', linestyle='--', linewidth=0.5)
plt.axvline(x=0, color='k', linestyle='--', linewidth=0.5)
plt.xlabel('x')
plt.ylabel('Binary Step(x)')
plt.title('Binary Step Activation Function')
plt.legend()
plt.grid()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

def linear(x):
    return x

# Generate x values from -10 to 10
x = np.linspace(-10, 10, 100)
y = linear(x)

# Plot the Linear function
plt.figure(figsize=(8, 5))
plt.plot(x, y, label='Linear Function', color='r')
plt.axhline(y=0, color='k', linestyle='--', linewidth=0.5)
plt.axvline(x=0, color='k', linestyle='--', linewidth=0.5)
plt.xlabel('x')
plt.ylabel('Linear(x)')
plt.title('Linear Activation Function')
plt.legend()
plt.grid()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

def tanh(x):
    return np.tanh(x)

def tanh_derivative(x):
    return 1 - np.tanh(x)**2

# Generate x values from -10 to 10
x = np.linspace(-10, 10, 100)
y = tanh(x)
y_derivative = tanh_derivative(x)

# Plot the Tanh function and its derivative
plt.figure(figsize=(8, 5))
plt.plot(x, y, label='Tanh Function', color='r')
plt.plot(x, y_derivative, label='Tanh Derivative', color='b', linestyle='dashed')
plt.axhline(y=0, color='k', linestyle='--', linewidth=0.5)
plt.axvline(x=0, color='k', linestyle='--', linewidth=0.5)
plt.xlabel('x')
plt.ylabel('Value')
plt.title('Tanh Activation Function and Its Derivative')
plt.legend()
plt.grid()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

def leaky_relu(x, alpha=0.01):
    return np.where(x > 0, x, alpha * x)

def leaky_relu_derivative(x, alpha=0.01):
    return np.where(x > 0, 1, alpha)

# Generate x values from -10 to 10
x = np.linspace(-10, 10, 100)
y = leaky_relu(x)
y_derivative = leaky_relu_derivative(x)

# Plot the Leaky ReLU function and its derivative
plt.figure(figsize=(8, 5))
plt.plot(x, y, label='Leaky ReLU Function', color='c')
plt.plot(x, y_derivative, label='Leaky ReLU Derivative', color='m', linestyle='dashed')
plt.axhline(y=0, color='k', linestyle='--', linewidth=0.5)
plt.axvline(x=0, color='k', linestyle='--', linewidth=0.5)
plt.xlabel('x')
plt.ylabel('Value')
plt.title('Leaky ReLU Activation Function and Its Derivative')
plt.legend()
plt.grid()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

def elu(x, alpha=1.0):
    return np.where(x > 0, x, alpha * (np.exp(x) - 1))

def elu_derivative(x, alpha=1.0):
    return np.where(x > 0, 1, elu(x, alpha) + alpha)

# Generate x values from -10 to 10
x = np.linspace(-10, 10, 100)
y = elu(x)
y_derivative = elu_derivative(x)

# Plot the ELU function and its derivative
plt.figure(figsize=(8, 5))
plt.plot(x, y, label='ELU Function', color='g')
plt.plot(x, y_derivative, label='ELU Derivative', color='r', linestyle='dashed')
plt.axhline(y=0, color='k', linestyle='--', linewidth=0.5)
plt.axvline(x=0, color='k', linestyle='--', linewidth=0.5)
plt.xlabel('x')
plt.ylabel('Value')
plt.title('ELU Activation Function and Its Derivative')
plt.legend()
plt.grid()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

def sigmoid(x):
    return 1 / (1 + np.exp(-x))

# Generate x values from -10 to 10
x = np.linspace(-10, 10, 100)
y = sigmoid(x)

# Plot the Sigmoid function
plt.figure(figsize=(8, 5))
plt.plot(x, y, label='Sigmoid Function', color='b')
plt.axhline(y=0, color='k', linestyle='--', linewidth=0.5)
plt.axhline(y=1, color='k', linestyle='--', linewidth=0.5)
plt.axvline(x=0, color='k', linestyle='--', linewidth=0.5)
plt.xlabel('x')
plt.ylabel('Sigmoid(x)')
plt.title('Sigmoid Activation Function')
plt.legend()
plt.grid()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

def relu(x):
    return np.maximum(0, x)

# Generate x values from -10 to 10
x = np.linspace(-10, 10, 100)
y = relu(x)

# Plot the ReLU function
plt.figure(figsize=(8, 5))
plt.plot(x, y, label='ReLU Function', color='g')
plt.axhline(y=0, color='k', linestyle='--', linewidth=0.5)
plt.axvline(x=0, color='k', linestyle='--', linewidth=0.5)
plt.xlabel('x')
plt.ylabel('ReLU(x)')
plt.title('ReLU Activation Function')
plt.legend()
plt.grid()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

def softmax(x):
    exp_x = np.exp(x - np.max(x))  # Subtract max for numerical stability
    return exp_x / exp_x.sum()

# Generate x values from -10 to 10
x = np.linspace(-5, 5, 10)
y = softmax(x)

# Plot the Softmax function
plt.figure(figsize=(8, 5))
plt.plot(x, y, marker='o', label='Softmax Function', color='m')
plt.xlabel('x')
plt.ylabel('Softmax(x)')
plt.title('Softmax Activation Function')
plt.legend()
plt.grid()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def tanh(x):
    return np.tanh(x)

def relu(x):
    return np.maximum(0, x)

def leaky_relu(x, alpha=0.01):
    return np.where(x > 0, x, alpha * x)

# Generate x values from -10 to 10
x = np.linspace(-10, 10, 100)

y_sigmoid = sigmoid(x)
y_tanh = tanh(x)
y_relu = relu(x)
y_leaky_relu = leaky_relu(x)

# Plot the activation functions
plt.figure(figsize=(10, 6))
plt.plot(x, y_sigmoid, label='Sigmoid', color='b')
plt.plot(x, y_tanh, label='Tanh', color='r')
plt.plot(x, y_relu, label='ReLU', color='g')
plt.plot(x, y_leaky_relu, label='Leaky ReLU', color='m')
plt.axhline(y=0, color='k', linestyle='--', linewidth=0.5)
plt.axvline(x=0, color='k', linestyle='--', linewidth=0.5)
plt.xlabel('x')
plt.ylabel('Activation Value')
plt.title('Activation Functions')
plt.legend()
plt.grid()
plt.show()
```
